# Inspecting the Zarr Structure

SHIVER provides three Zarr stores for each of Greenland and Antarctica. Each Zarr is optimized for different types of analysis. Before we go on to explore those types of analysis, it's useful to first inspect the structure of these Zarr stores. 

To begin with, let's examine the structure of `antarctica_multisource_velocity_spatial.zarr`, which is chunked shallowly in time to allow large areas of data to be loaded into memory with minimal resources. The Greenland equivalent uses the same chunk definition. 

These `spatial` Zarrs are formatted as an **OME-Zarr**, meaning they contain an image pyramid of multiple resolution layers (groups) to optimize broad spatial queries. We can view the available resolution layers and then use `xarray` to load the desired group. Group "0" is always the native (highest) resolution, while higher group numbers represent progressively downsampled resolutions.


```python
import zarr
import xarray as xr

url = "https://data.source.coop/uos-shiver/antarctica/antarctica_multisource_velocity_spatial.zarr"

# View the OME-Zarr multiscale groups (resolution levels)
root = zarr.open_group(url, mode='r')
print("Resolution levels:", list(root.group_keys()))

# Open the highest resolution layer (group "0") to inspect the dataset structure
ds_spatial_hr = xr.open_zarr(url, group="0", consolidated=True)
ds_spatial_hr
```

We can similarly examine the structure of the time-series optimized Zarr store for Antarctica. Since this is a standard Zarr (rather than an OME-Zarr, there is no need to specify the group level.

```python
url = "https://data.source.coop/uos-shiver/antarctica/antarctica_multisource_velocity_timeseries.zarr"

# Open the highest resolution layer (group "0") to inspect the dataset structure
ds_timeseries = xr.open_zarr(url, consolidated=True)
ds_timeseries
```

And we can view the structure of the `cubed` Zarr, like this.

```python
url = "https://data.source.coop/uos-shiver/antarctica/antarctica_multisource_velocity_cubed.zarr"

# Open the highest resolution layer (group "0") to inspect the dataset structure
ds_cubed = xr.open_zarr(url, consolidated=True)
ds_cubed
```
