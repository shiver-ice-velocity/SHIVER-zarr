# Generating citations for Contributing Datasets

The SHIVER unified datasets are a compilation of many published ice velocity datasets. Anytime you use output from SHIVER, you should cite the contributing datasets that your query intersected, as well as the unified dataset. 

If you extract a timeseries of ice velocity from a given location, your query could intersect up to 17 contributing datasets. This notebook describes how to automatically generate citations for those datasets.

To begin with, let's repeat the timeseries extraction shown in the previous notebook and pull the data sources for that timeseries.

```python
import xarray as xr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pyproj import Transformer

greenland_url = "https://data.source.coop/uos-shiver/greenland/greenland_multisource_velocity_timeseries.zarr"

# Convert geographic coordinates (Lon/Lat) to EPSG:3413 (X/Y)
transformer = Transformer.from_crs("EPSG:4326", "EPSG:3413", always_xy=True)
lon, lat = -50.1874, 68.8267
x_center, y_center = transformer.transform(lon, lat)

# Create a 500m spatial buffer bounding box
buffer = 500
xmin, xmax = x_center - buffer, x_center + buffer
ymin, ymax = y_center - buffer, y_center + buffer

# Load the timeseries-optimized Zarr store lazily
ds = xr.open_zarr(greenland_url, consolidated=True).sortby('time')

# Handle Y-axis orientation for slicing
y_slice = slice(ymax, ymin) if ds.y[0] > ds.y[-1] else slice(ymin, ymax)

# Extract data lazily and compute the spatial median
subset = ds.sel(x=slice(xmin, xmax), y=y_slice)
spatial_median = subset['speed'].median(dim=['x', 'y']).compute()

# Convert the in-memory array to a Pandas DataFrame
df = spatial_median.to_dataframe()
df = df.dropna(subset=['speed']).sort_index()

# Calculate a 24-day rolling mean
df['rolling_mean'] = df['speed'].rolling('24D').mean()

# Filter for plotting (2016 onwards)
df_plot = df[df.index >= '2016-01-01']

# Make and apply a mask of the times that fall in the plot window
valid_times = df_plot.index
time_mask = subset['time'].isin(valid_times)
raw_sources = subset['data_source'].where(time_mask, drop=True).values.flatten()

# Create unique list of data sources that we need to cite
unique_sources = list(np.unique([str(s) for s in raw_sources if pd.notna(s) and str(s).lower() not in ['nan', 'none', '']]))
```

Now we use a citation generation function, `generate_citation_table`, that accepts this list of sources. Each SHIVER store records its release in `ds.attrs["dataset_version"]`, which matches a version of the Zenodo record, so we pass that in too and the SHIVER citation names the exact release you used. 

> **Note: This relies on a hard-coded set of citations provided separately (`utils/citations.py`). Whilst we will endeavour to maintain this utility, it is possible for dataset citations to change as versions increment, which will not instantly be reflected in this utility. As the investigator, you are responsible for providing the correct citation(s) for the data used in your analysis.** 

```python
from IPython.display import display
import pandas as pd
from utils.citations import CITATIONS_CONFIG, shiver_citations

def generate_citation_table(sources_list: list, region: str, dataset_version=None):
    """
    Takes a list of data sources and returns a styled Pandas DataFrame
    containing the relevant citations, grouped with one row per source.
    """
    table_data = []
    
    # 1. Always append the SHIVER foundational citations
    # Join multiple citations with a double newline
    combined_shiver = "\n\n".join(shiver_citations(dataset_version))
    table_data.append({"Data Source": "SHIVER (Tool/Method)", "Citation": combined_shiver})
        
    # 2. Add the dynamic data citations
    region_key = region.lower()
    region_citations = CITATIONS_CONFIG.get(region_key, {})
    
    for source in sources_list:
        if source in region_citations:
            citations = region_citations[source]
            if isinstance(citations, str):
                citations = [citations]

            ## One could filter the ESA_CCI_Sentinel-2 citations here, if desired
            #if source == "ESA_CCI_Sentinel-2":
            #    if len(citations) >= 3:
            #        citations = [citations[2]]  # Index 2 corresponds to the 3rd item
            #    else:
            #        print(f"Warning: 'ESA_CCI_Sentinel-2' has fewer than 3 citations available.")
            
            # Join multiple citations for a single source with a double newline
            combined_citations = "\n\n".join(citations)
            table_data.append({"Data Source": source, "Citation": combined_citations})
            
        else:
            table_data.append({
                "Data Source": source, 
                "Citation": f"WARNING: Missing citation for source '{source}' in region '{region}'"
            })
            
    # 3. Format as a readable table
    df_citations = pd.DataFrame(table_data)
    
    # Apply styling
    styled_df = df_citations.style.set_properties(**{
        'text-align': 'left',
        'white-space': 'pre-wrap', 
        'vertical-align': 'top',
        'padding': '10px'
    }).set_table_styles([
        {'selector': 'th', 'props': [('text-align', 'left'), ('font-size', '14px')]}
    ]).hide(axis="index")
    
    return styled_df
```

And we simply pass our list of `unique_sources` to `generate_citation_table`

```python
region = "greenland"

# Display the unique sources found for a quick sanity check
print(f"Data sources found from 2016 onwards: {', '.join(unique_sources)}")
print(f"SHIVER dataset version: {ds.attrs.get('dataset_version')}\n")

# Generate and display the table
citation_table = generate_citation_table(unique_sources, region, ds.attrs.get("dataset_version"))
display(citation_table)
```

Of course you may wish to use a more compact format for your citation table. 

Then you can generate a BibTeX file, which you can import into your choice of reference manager, like this:

```python
from utils.citations_bib_library import CITATIONS_DB, GREENLAND_SOURCE_MAPPING

def generate_structured_bib(required_sources):
    # 1. Ensure SHIVER is always included
    sources_to_cite = set(required_sources)
    sources_to_cite.add("SHIVER")
    
    # 2. Collect unique citation keys (using a set automatically removes duplicates)
    citation_keys = set()
    for source in sources_to_cite:
        if source in GREENLAND_SOURCE_MAPPING:
            citation_keys.update(GREENLAND_SOURCE_MAPPING[source])
            
    # 3. Generate BibTeX entries
    bib_entries = []
    for key in citation_keys:
        data = dict(CITATIONS_DB[key])
        if key == "davison_shiver_data":
            # Name the exact release of the store the data came from
            data["version"] = ds.attrs.get("dataset_version")
        
        # Build the fields dynamically based on what exists in the dictionary
        fields = []
        for field_name in ["author", "year", "title", "journal", "volume", "pages", "publisher", "howpublished", "version", "note", "doi", "url"]:
            if field_name in data and data[field_name]:
                fields.append(f"  {field_name} = {{{data[field_name]}}}")
                
        fields_str = ",\n".join(fields)
        
        bib_entry = (
            f"@{data['type']}{{{key},\n"
            f"{fields_str}\n"
            f"}}\n\n"
        )
        bib_entries.append(bib_entry)
        
    return bib_entries

# --- Execution ---
bib_content = generate_structured_bib(unique_sources)

with open("citations.bib", "w", encoding="utf-8") as f:
    f.writelines(bib_content)

print(f"Successfully generated {len(bib_content)} unique BibTeX entries.")
```

> _Note: The ESA_CCI_Sentinel-1, ESA_CCI_Sentinel-2 and ESA_CCI_ERS1-2_Envisat over selected glacier sites in Greenland require separate citations for each glacier site. Our dataset compilation method combines all of those glacier sites into one Contributing Dataset. This automated citation generator will therefore provide the citation information for all glacier sites even if only one is intersected (as was the case in this example). You should remove the unnecessary citations when incorporating them into your work._
